```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
# import df
file_path = "aircraft_m.csv"
df = pd.read_csv(file_path)
```

```python
df.head()
```

```python
#split into dependent and independent
X = df.drop(columns=["Fuel_Efficiency_L_per_km"])
y = df["Fuel_Efficiency_L_per_km"]
```

```python
#Split data into training (70%) and testing (30%)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
#train model using  Multiple Linear Regression Model
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
# Make predictions
y_train_pred = model.predict(X_train)  
y_test_pred = model.predict(X_test)    
```

```python
# Evaluate the model
mse_train = mean_squared_error(y_train, y_train_pred)
mse_test=mean_squared_error(y_test, y_test_pred)
mae_train = mean_absolute_error(y_train, y_train_pred)
mae_test =mean_absolute_error(y_test, y_test_pred)
r2_train = r2_score(y_train, y_train_pred)
r2_test = r2_score(y_test, y_test_pred)
```

```python
#print evaluation metrics

print("Model Evaluation Scores")
print(f"Training Set: MSE = {mse_train}, MAE = {mae_train}, R2 = {r2_train}")
print(f"Testing Set:  MSE = {mse_test}, MAE = {mae_test}, R2 = {r2_test}")
```

```python
coefficients = np.abs(model.coef_)
feature_names = X.columns
```

```python
sorted_indices = np.argsort(coefficients)[::-1]
sorted_features = feature_names[sorted_indices]
sorted_coefficients = coefficients[sorted_indices]
```

```python
print("Independent Variables Ranked by Importance (highest to lowest)")
for feature, coef, in zip(sorted_features, sorted_coefficients):
    print(f"{feature}: {coef}")
    
print("-"*25)

print("Actual vs Predicted:")
for actual, predicted in zip(y_test[:10], y_test_pred[:10]):
    print(f"Actual: {actual}, Predicted: {predicted}")
```
