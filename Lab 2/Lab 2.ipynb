## Lab 2: Basic Pandas Operations

#### Objective:
In this lab, you will learn how to import a CSV file into a pandas DataFrame and perform basic operations such as viewing the first few rows, the last few rows, and generating summary statistics of the DataFrame.

#### Step-by-Step Instructions:


#### 1. Import the pandas Library

First, we need to import the pandas library. Pandas is a powerful library for data manipulation and analysis in Python.

```python
import pandas as pd
```

#### 2. Load a CSV File into a DataFrame

Let's load a sample CSV file into a DataFrame. For this demo, we'll use a CSV file named `students_scores.csv`. Make sure this file is in your working directory or provide the full path to the file.

```python
df = pd.read_csv('students_scores.csv')
```

#### 3. View the First Few Rows of the DataFrame

To get a quick look at the first few rows of the DataFrame, we use the `head()` method. By default, it shows the first 5 rows, but you can specify the number of rows to display.

```python
print("First 5 rows of the DataFrame:")
print(df.head())
```

#### 4. View the Last Few Rows of the DataFrame

Similarly, to see the last few rows, we use the `tail()` method. It also shows 5 rows by default, but you can change this number as well.

```python
print("Last 5 rows of the DataFrame:")
print(df.tail())
```

#### 5. Generate Summary Statistics

The `describe()` method provides a quick overview of the numerical columns in the DataFrame, including count, mean, standard deviation, min, and max values, as well as the 25th, 50th (median), and 75th percentiles.

```python
print("Summary statistics of the DataFrame:")
print(df.describe())
```
