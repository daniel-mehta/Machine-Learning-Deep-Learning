```python
import pandas as pd  # Importing pandas for data manipulation
from sklearn.model_selection import train_test_split  # Importing train_test_split from sklearn
```

```python
# Let's start by loading our dataset into a pandas DataFrame.
car_data = pd.read_csv('fuel_efficiency_data.csv')  # Reading the CSV file into a DataFrame
```

```python
# Separating the independent variable (Engine Size in Liters) and dependent variable (Fuel Efficiency in MPG).
indp_vars = car_data[['Engine_Size_Liters']]  # Selecting the 'Engine_Size_Liters' column as independent variable
dep_var = car_data['Fuel_Efficiency_MPG']  # Selecting the 'Fuel_Efficiency_MPG' column as dependent variable

```

```python
# Splitting the data into training and testing sets.
train_x, test_x, train_y, test_y = train_test_split(indp_vars, dep_var, test_size=0.1, random_state=42)  # Splitting the data with 80% for training and 20% for testing
test_x
```

```python
train_x, test_x, train_y, test_y = train_test_split(indp_vars, dep_var, test_size=0.1, random_state=101)  # Splitting the data with 80% for training and 20% for testing
test_x
```
