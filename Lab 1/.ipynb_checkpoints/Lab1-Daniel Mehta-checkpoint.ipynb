#### Control Structures

##### Create a loop that prints numbers from 1 to 10.

```python
for x in range(10):
    print(x+1)
```

##### Create a loop that prints even numbers between 1 to 10.

```python
for x in range(10):
    x+=1
    if (x % 2) == 0:
        print(x)
```

##### Use an if-else structure to check if a number is odd or even

```python
num = float(input("Enter Number: "))
if (num % 2) == 0:
    print(f"{num} is even")
else:
    print(f"{num} is odd")
```

#### Functions

##### Write a function that takes a number as an argument and returns its square.

```python
def square(num):
    return num * num

square(5)
```

##### Write a function that takes a string as an argument and returns its reverse.

```python
def reverse_string(string):
    return string[::-1]

reverse_string("Hello World!")
```

##### Write a Python function named second_largest that takes a list of integers and returns the second-largest number in the list. Assume the list has at least two distinct elements.

```python
def second_largest(list):
    largest = second_largest = float('-inf')
    
    for n in list:
        if n > largest:
            second_largest = largest
            largest = n
        elif n > second_largest and largest != n:
            second_largest = n
    return second_largest

second_largest([5,2,8,1,7])
```

##### Write a function named accumulate_squares that takes an integer n and returns the sum of the squares of all numbers from 1 to n that are not divisible by 3.

```python
def accumulate_squares(n):
    return sum(i**2 for i in range(1, n+1 ) if i% 3 != 0)

accumulate_squares(5)
```
